```python
import os

file_name = 'final_tickets_tenge_df.csv'
if not os.path.exists(file_name):
    print('Файл курса не найден, создаю временный датафрейм...')
    dates = pd.date_range(start='2024-01-01', end='2024-12-31')
    tenge = pd.DataFrame({'data': dates, 'kzt_rate_per_100': 19.5})
else:
    tenge = pd.read_csv(file_name)
    tenge = tenge.rename(columns={'curs': 'kzt_rate_per_100'})

tenge['data'] = pd.to_datetime(tenge['data'])
tenge = tenge.rename(columns={'data': 'order_dt'})
tenge = tenge[['order_dt', 'kzt_rate_per_100']]
print(tenge.head())
```

```python
df['order_dt'] = pd.to_datetime(df['order_dt'], unit='s').dt.normalize()
df['order_ts'] = pd.to_datetime(df['order_ts'], unit='s')

# Слияние с курсом валют
df = df.merge(tenge, on='order_dt', how='left')
print('Пропуски после merge курса:', df['kzt_rate_per_100'].isna().sum())
```

```python
df['revenue_rub'] = np.where(
    df['currency_code'] == 'kzt',
    df['revenue'] / 100 * df['kzt_rate_per_100'],
    df['revenue']
)
print(df[['currency_code', 'revenue', 'kzt_rate_per_100', 'revenue_rub']].head(10))
```

### 3. Создание профиля пользователя

В будущем отдел маркетинга планирует создать модель для прогнозирования возврата пользователей. Поэтому сейчас они просят вас построить агрегированные признаки, описывающие поведение и профиль каждого пользователя.

**Задача 3.1.** Постройте профиль пользователя — для каждого пользователя найдите:

- дату первого и последнего заказа;
- устройство, с которого был сделан первый заказ;
- регион, в котором был сделан первый заказ;
- билетного партнёра, к которому обращались при первом заказе;
- жанр первого посещённого мероприятия (используйте поле `event_type_main`);
- общее количество заказов;
- средняя выручка с одного заказа в рублях;
- среднее количество билетов в заказе;
- среднее время между заказами.

После этого добавьте два бинарных признака:

- `is_two` — совершил ли пользователь 2 и более заказа;
- `is_five` — совершил ли пользователь 5 и более заказов.

**Рекомендация:** перед тем как строить профиль, отсортируйте данные по времени совершения заказа.

---


```python
df = df.sort_values(['user_id', 'order_ts']).reset_index(drop=True)

first_orders = df.groupby('user_id').first().reset_index()[
    ['user_id', 'order_dt', 'device_type_canonical', 'region_name', 'service_name', 'event_type_main']
].rename(columns={
    'order_dt': 'first_order_dt',
    'device_type_canonical': 'first_device',
    'region_name': 'first_region',
    'service_name': 'first_service',
    'event_type_main': 'first_event_type'
})

last_orders = df.groupby('user_id')['order_dt'].max().reset_index().rename(columns={'order_dt': 'last_order_dt'})
```

```python
agg = df.groupby('user_id').agg(
    total_orders=('order_id', 'nunique'),
    avg_revenue_rub=('revenue_rub', 'mean'),
    avg_tickets_count=('tickets_count', 'mean'),
    avg_days_between=('days_since_prev', 'mean')
).reset_index()

user_profile = first_orders.merge(last_orders, on='user_id').merge(agg, on='user_id')

user_profile['is_two'] = (user_profile['total_orders'] >= 2).astype(int)
user_profile['is_five'] = (user_profile['total_orders'] >= 5).astype(int)

user_profile.head()
```

```python
print('Размер профиля:', user_profile.shape)
print(user_profile[['total_orders', 'avg_revenue_rub', 'avg_tickets_count', 'avg_days_between']].describe().T)
```

Сегмент — это группа пользователей, объединённых определённым признаком, то есть объединённые принадлежностью к категории. Например, все клиенты, сделавшие первый заказ с мобильного телефона, — это сегмент.
- Ответьте на вопрос: равномерно ли распределены пользователи по сегментам или есть выраженные «точки входа» — сегменты с наибольшим числом пользователей?

```python
def segment_share(series, name):
    counts = series.value_counts()
    shares = series.value_counts(normalize=True)
    result = pd.DataFrame({'users': counts, 'share': shares})
    result.index.name = name
    return result

seg_event = segment_share(user_profile['first_event_type'], 'first_event_type')
seg_device = segment_share(user_profile['first_device'], 'first_device')
seg_region = segment_share(user_profile['first_region'], 'first_region')
seg_service = segment_share(user_profile['first_service'], 'first_service')

print('По типу мероприятия:')
print(seg_event.head(5))
print('\nПо устройству:')
print(seg_device)
```

**Задача 4.1.2.** Проанализируйте возвраты пользователей:

- Для каждого сегмента вычислите долю пользователей, совершивших два и более заказа.
- Визуализируйте результат подходящим графиком. Если сегментов слишком много, то поместите на график только 10 сегментов с наибольшим количеством пользователей. Такое возможно с сегментами по региону и по билетному оператору.
- Ответьте на вопросы:
    - Какие сегменты пользователей чаще возвращаются на Яндекс Афишу?
    - Наблюдаются ли успешные «точки входа» — такие сегменты, в которых пользователи чаще совершают повторный заказ, чем в среднем по выборке?

При интерпретации результатов учитывайте размер сегментов: если в сегменте мало пользователей (например, десятки), то доли могут быть нестабильными и недостоверными, то есть показывать широкую вариацию значений.

---


```python
def return_rate_by(df_, col, top_n=None):
    g = df_.groupby(col).agg(users=('user_id', 'nunique'), returned=('is_two', 'sum'))
    g['return_rate'] = g['returned'] / g['users']
    g = g.sort_values('users', ascending=False)
    return g.head(top_n) if top_n else g

rr_event = return_rate_by(user_profile, 'first_event_type')
rr_device = return_rate_by(user_profile, 'first_device')
rr_region = return_rate_by(user_profile, 'first_region', top_n=10)
rr_service = return_rate_by(user_profile, 'first_service', top_n=10)

overall_return = user_profile['is_two'].mean()
print(f'Средняя доля возврата по выборке: {overall_return:.2%}')
print('\nПо региону (топ-10):')
print(rr_region)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
rr_event.plot(kind='bar', y='return_rate', ax=axes[0, 0], title='Возврат по мероприятию', legend=False)
rr_device.plot(kind='bar', y='return_rate', ax=axes[0, 1], title='Возврат по устройству', legend=False)
rr_region.plot(kind='bar', y='return_rate', ax=axes[1, 0], title='Возврат по региону (топ-10)', legend=False)
rr_service.plot(kind='bar', y='return_rate', ax=axes[1, 1], title='Возврат по оператору (топ-10)', legend=False)

for ax in axes.flat:
    ax.axhline(overall_return, color='red', linestyle='--', label='среднее')
    ax.tick_params(axis='x', rotation=45)
    ax.legend()
plt.tight_layout()
plt.show()
```

Чтобы этого избежать, выделите сегменты пользователей по полю `total_orders`, а затем повторите корреляционный анализ. Выделите такие сегменты:

```python
import phik

# Для общей корреляции исключаем avg_days_between, так как там есть NaN
profile_cols = [
    'first_device', 'first_region', 'first_service', 'first_event_type',
    'total_orders', 'avg_revenue_rub', 'avg_tickets_count'
]
corr_data = user_profile[profile_cols].copy()
interval_cols = ['total_orders', 'avg_revenue_rub', 'avg_tickets_count']

print("Корреляция phi_k с total_orders (по всем пользователям):")
phik_matrix = corr_data.phik_matrix(interval_cols=interval_cols)
print(phik_matrix['total_orders'].sort_values(ascending=False))

# Корреляция только для вернувшихся пользователей, включая avg_days_between
returned_users_df = user_profile[user_profile['total_orders'] >= 2]
profile_cols_returned = [
    'first_device', 'first_region', 'first_service', 'first_event_type',
    'avg_revenue_rub', 'avg_tickets_count', 'avg_days_between'
]
corr_data_returned = returned_users_df[profile_cols_returned].copy()
interval_cols_returned = ['avg_revenue_rub', 'avg_tickets_count', 'avg_days_between']

print("\nКорреляция phi_k между признаками (по вернувшимся пользователям):")
phik_matrix_returned = corr_data_returned.phik_matrix(interval_cols=interval_cols_returned)
print(phik_matrix_returned)
```

```python
def order_segment(x):
    if x == 1: return '1 заказ'
    elif x <= 4: return '2-4 заказа'
    else: return '5+ заказов'

user_profile['orders_segment'] = user_profile['total_orders'].apply(order_segment)
corr_by_seg = {}

for seg_name, seg_df in user_profile.groupby('orders_segment'):
    if seg_name == '1 заказ':
        cols = ['first_device', 'first_region', 'first_service', 'first_event_type', 'avg_revenue_rub', 'avg_tickets_count']
        interval_c = ['avg_revenue_rub', 'avg_tickets_count']
    else:
        cols = ['first_device', 'first_region', 'first_service', 'first_event_type', 'avg_revenue_rub', 'avg_tickets_count', 'avg_days_between']
        interval_c = ['avg_revenue_rub', 'avg_tickets_count', 'avg_days_between']
        
    m = seg_df[cols].phik_matrix(interval_cols=interval_c)
    corr_by_seg[seg_name] = m
```

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
for ax, (seg_name, m) in zip(axes, corr_by_seg.items()):
    sns.heatmap(m, annot=True, fmt='.2f', cmap='coolwarm', ax=ax, vmin=0, vmax=1)
    ax.set_title(seg_name)
plt.tight_layout()
plt.show()
```

Заметной связи дня недели первого заказа с возвратом нет. Доли возврата по дням отличаются слабо, в пределах нескольких процентных пунктов. Это означает, что день недели, в который пользователь впервые совершил покупку, не стоит использовать как самостоятельный признак для таргетинга.
